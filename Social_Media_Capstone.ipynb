```python

import pandas as pd

data = {
    "Platform": [
        "Instagram", "Facebook", "YouTube", "Instagram", "Facebook",
        "YouTube", "Instagram", "Facebook", "YouTube", "Instagram",
        "Facebook", "YouTube", "Instagram", "Facebook", "YouTube",
        "Instagram", "Facebook", "YouTube", "Instagram", "Facebook",
        "YouTube", "Instagram", "Facebook", "YouTube", "Instagram",
        "Facebook", "YouTube", "Instagram", "Facebook", "YouTube"
    ],

    "Likes": [
        500, 300, 600, 800, 400, 700, 650, 350, 900, 750,
        450, 800, 550, 500, 650, 900, 600, 750, 700, 550,
        850, 1000, 650, 950, 850, 700, 1000, 950, 750, 1100
    ],

    "Shares": [
        120, 80, 150, 200, 100, 180, 140, 90, 220, 170,
        110, 190, 130, 120, 160, 210, 140, 180, 150, 130,
        200, 250, 160, 230, 190, 170, 240, 220, 180, 260
    ],

    "Comments": [
        45, 30, 60, 90, 40, 75, 55, 35, 100, 80,
        45, 85, 50, 45, 65, 95, 55, 80, 70, 50,
        90, 110, 60, 105, 85, 65, 115, 100, 75, 120
    ]
}

df = pd.DataFrame(data)

print("Dataset Created Successfully!")
print(df)
```

```python

from IPython.display import display

pd.set_option('display.max_rows', 50)
pd.set_option('display.max_columns', None)

display(df)
```

```python

# 1. Dataset ka size check karna
print("Dataset Shape:")
print(df.shape)

# 2. Dataset ki information
print("\nDataset Information:")
df.info()

# 3. Statistical Summary
print("\nDataset Description:")
display(df.describe())

# 4. Platform count
print("\nPlatform Wise Count:")
print(df["Platform"].value_counts())
```

```python

# Correlation Analysis

correlation = df[["Likes", "Shares", "Comments"]].corr()

print("Correlation Between Variables:")
display(correlation)
```

```python

import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 5))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Social Media Engagement Correlation")
plt.show()
```

```python

# Platform Wise GroupBy Analysis

platform_analysis = df.groupby("Platform")[["Likes", "Shares", "Comments"]].mean()

print("Platform Wise Average Engagement:")
display(platform_analysis)
```

```python

import matplotlib.pyplot as plt

platform_analysis.plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Average Social Media Engagement by Platform")
plt.xlabel("Social Media Platform")
plt.ylabel("Average Engagement")
plt.xticks(rotation=0)
plt.legend(title="Engagement Type")
plt.tight_layout()
plt.show()
```

```python

df["Post_Date"] = pd.date_range(
    start="2026-01-01",
    periods=len(df),
    freq="D"
)

print("Date Column Added!")
display(df.head())
```

```python

import matplotlib.pyplot as plt

platform_counts = df["Platform"].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    platform_counts,
    labels=platform_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Social Media Platform Distribution")
plt.show()
```

```python

plt.figure(figsize=(10, 5))

plt.plot(
    df["Post_Date"],
    df["Likes"],
    marker="o"
)

plt.title("Date-wise Social Media Likes")
plt.xlabel("Post Date")
plt.ylabel("Number of Likes")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python

import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Platform wise analysis
platform_analysis = df.groupby("Platform")[
    ["Likes", "Shares", "Comments"]
].mean()

# Create dashboard
fig = make_subplots(
    rows=2,
    cols=2,
    subplot_titles=(
        "Average Engagement by Platform",
        "Date-wise Likes",
        "Platform Distribution",
        "Likes vs Shares"
    ),
    specs=[
        [{"type": "bar"}, {"type": "scatter"}],
        [{"type": "domain"}, {"type": "scatter"}]
    ]
)

# 1. Bar Chart
for column in ["Likes", "Shares", "Comments"]:
    fig.add_trace(
        go.Bar(
            x=platform_analysis.index,
            y=platform_analysis[column],
            name=column
        ),
        row=1, col=1
    )

# 2. Line Chart
for platform in df["Platform"].unique():
    temp = df[df["Platform"] == platform]

    fig.add_trace(
        go.Scatter(
            x=temp["Post_Date"],
            y=temp["Likes"],
            mode="lines+markers",
            name=platform
        ),
        row=1, col=2
    )

# 3. Pie Chart
fig.add_trace(
    go.Pie(
        labels=df["Platform"].value_counts().index,
        values=df["Platform"].value_counts().values,
        name="Platforms"
    ),
    row=2, col=1
)

# 4. Scatter Chart
fig.add_trace(
    go.Scatter(
        x=df["Likes"],
        y=df["Shares"],
        mode="markers",
        name="Likes vs Shares"
    ),
    row=2, col=2
)

fig.update_layout(
    title="Social Media Engagement Analytics Dashboard",
    height=850,
    barmode="group",
    template="plotly_white"
)

fig.show()
```

```python

# Export dashboard as HTML

fig.write_html(
    "Social_Media_Dashboard.html",
    include_plotlyjs=True
)

print("Dashboard Exported Successfully!")
```
